```python
null_counts = df.isnull().sum()
print(null_counts[null_counts > 0])
```

```python
df.duplicated().sum()
df.duplicated(subset=["Order.ID", "Customer.ID"]).sum()
```

```python
df.rename(columns={df.columns[7]: "record_count"}, inplace=True)
```

```python
df.columns=(df.columns.str.strip().str.lower().str.replace(".","_").str.replace(" ","_"))
```

```python
print(df.columns.tolist())
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
df["ship_date"] = pd.to_datetime(df["ship_date"], errors="coerce")
df[["order_date", "ship_date"]].info()
```

```python
df = df.drop_duplicates(subset=["order_id", "customer_id"])
df.duplicated().sum()
```

```python
df["shipping_cost"].fillna(df["shipping_cost"].median(), inplace=True)

```

```python

df = df.dropna(subset=["customer_id"])
```

```python
numeric_cols = ["sales", "profit", "quantity", "discount", "shipping_cost"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["year_month"] = df["order_date"].dt.to_period("M").astype(str)
```

```python
df.to_csv("../data/superstore_clean.csv",index=False)
print("Cleaned csv saved!!")
```
